```python
from geopy.geocoders.pickpoint import Nominatim
import pandas as pd
from geopy.geocoders import Nominatim
from geopy.exc import GeocoderTimedOut, GeocoderUnavailable
import time
```

```python
data = pd.read_excel('/content/drive/MyDrive/Analysis/Coordonnées_OTB.xlsx')
```

```python
# Initialisation du géocodeur
geolocator = Nominatim(
    user_agent="geo_app",
    timeout=10
)

# Fonction de recherche d'adresse
def get_address(lat, lon):

    # Corriger automatiquement si les colonnes sont inversées
    if abs(lat) > 90 and abs(lon) <= 90:
        lat, lon = lon, lat

    for _ in range(3):  # 3 tentatives

        try:
            location = geolocator.reverse(
                (lat, lon),
                language="fr",
                exactly_one=True
            )

            return location.address if location else None

        except (GeocoderTimedOut, GeocoderUnavailable):
            time.sleep(5)

        except Exception as e:
            print(f"Erreur : {e}")
            return None

    return None


# Recherche des adresses
adresses = []

for i, row in data.iterrows():

    adresse = get_address(
        row["Latitude"],
        row["Longitude"]
    )

    adresses.append(adresse)

    if i % 100 == 0:
        print(f"{i} lignes traitées")

    time.sleep(1)  # respecter les limites Nominatim


# Ajout au dataframe
data["Adresse"] = adresses

# Sauvegarde
data.to_excel(
    "coordonnees_avec_adresses.xlsx",
    index=False
)

print("Terminé")
print(f"{len(data)} lignes traitées")
```

```python
def split_address(adresse):
    if pd.isna(adresse):
        return pd.Series([None, None, None, None])

    parts = [p.strip() for p in str(adresse).split(",")]

    pays = parts[-1] if len(parts) >= 1 else None
    ville = parts[-2] if len(parts) >= 2 else None
    commune = parts[-3] if len(parts) >= 3 else None
    quartier = parts[-4] if len(parts) >= 4 else None

    return pd.Series([quartier, commune, ville, pays])

data[['Quartier','Commune','Ville','Pays']] = data['Adresse'].apply(split_address)
```
